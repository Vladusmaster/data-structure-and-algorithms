```python
# ==================== UFC FIGHT PREDICTOR - COMPLETE VERSION ====================
# Включает 6 графиков и тестирование 5 оптимизаторов

import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam, SGD, RMSprop, Adagrad, Nadam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.regularizers import l2
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, roc_curve, auc
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print("="*70)
print("🥊 UFC FIGHT PREDICTOR - ПОЛНАЯ ВЕРСИЯ")
print("6 графиков + 5 оптимизаторов")
print("="*70)

# ==================== 1. ЗАГРУЗКА ДАННЫХ ====================
print("\n1. 📥 Загрузка данных...")

fighters = pd.read_csv("fighters.csv")
fights = pd.read_csv("Fights.csv")

print(f"   Бойцы: {fighters.shape}")
print(f"   Бои: {fights.shape}")

# ==================== 2. ПОДГОТОВКА ДАННЫХ ====================
print("\n2. 🧹 Подготовка данных...")

def prepare_pre_fight_features(fights_df, fighters_df):
    fighters_processed = fighters_df.copy()

    def convert_height(height):
        if pd.isna(height):
            return np.nan
        try:
            if isinstance(height, str) and "'" in height:
                feet, inches = height.split("'")
                feet = float(feet)
                inches = float(inches.replace('"', '').strip())
                return feet * 30.48 + inches * 2.54
            return float(height)
        except:
            return np.nan

    fighters_processed['height_cm'] = fighters_processed['Ht.'].apply(convert_height)
    fighters_processed['reach_cm'] = fighters_processed['Reach'] * 2.54

    fighters_processed['height_cm'] = fighters_processed['height_cm'].fillna(fighters_processed['height_cm'].median())
    fighters_processed['reach_cm'] = fighters_processed['reach_cm'].fillna(fighters_processed['reach_cm'].median())
    fighters_processed['W'] = fighters_processed['W'].fillna(0)
    fighters_processed['L'] = fighters_processed['L'].fillna(0)
    fighters_processed['D'] = fighters_processed['D'].fillna(0)

    fighter_chars = {}
    for idx, row in fighters_processed.iterrows():
        name = row['Full Name']
        total_fights = row['W'] + row['L'] + row['D']
        win_pct = row['W'] / (total_fights + 1e-10)

        fighter_chars[name] = {
            'height': row['height_cm'],
            'reach': row['reach_cm'],
            'wins': row['W'],
            'losses': row['L'],
            'draws': row['D'],
            'total_fights': total_fights,
            'win_pct': win_pct,
            'belt': 1 if row['Belt'] else 0
        }

    features_list = []
    targets_list = []

    print(f"   Обработка {len(fights_df)} боев...")

    for idx, row in fights_df.iterrows():
        fighter1 = row['Fighter_1']
        fighter2 = row['Fighter_2']

        f1_chars = fighter_chars.get(fighter1, {})
        f2_chars = fighter_chars.get(fighter2, {})

        if not f1_chars or not f2_chars:
            continue

        features = [
            f1_chars.get('height', 0) - f2_chars.get('height', 0),
            f1_chars.get('reach', 0) - f2_chars.get('reach', 0),
            f1_chars.get('wins', 0) - f2_chars.get('wins', 0),
            f1_chars.get('losses', 0) - f2_chars.get('losses', 0),
            f1_chars.get('win_pct', 0) - f2_chars.get('win_pct', 0),
            f1_chars.get('total_fights', 0) - f2_chars.get('total_fights', 0),
            f1_chars.get('height', 0),
            f1_chars.get('reach', 0),
            f1_chars.get('wins', 0),
            f1_chars.get('win_pct', 0),
            f1_chars.get('total_fights', 0),
            f1_chars.get('belt', 0),
            f2_chars.get('height', 0),
            f2_chars.get('reach', 0),
            f2_chars.get('wins', 0),
            f2_chars.get('win_pct', 0),
            f2_chars.get('total_fights', 0),
            f2_chars.get('belt', 0),
        ]

        if not any(np.isnan(features)):
            features_list.append(features)
            if 'Result_1' in row:
                target = 1 if row['Result_1'] == 'W' else 0
                targets_list.append(target)

    X = np.array(features_list)
    y = np.array(targets_list)

    print(f"   ✅ Создано {X.shape[0]} примеров с {X.shape[1]} признаками")
    print(f"   📊 Распределение классов: {sum(y==1)} / {sum(y==0)}")

    return X, y

X, y = prepare_pre_fight_features(fights, fighters)

# ==================== 3. ПОДГОТОВКА ВЫБОРОК ====================
print("\n3. 📊 Подготовка выборок...")

X = np.where(np.isinf(X), 100, X)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42, stratify=y
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.15, random_state=42, stratify=y_train
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print(f"   Обучающая: {X_train.shape}")
print(f"   Валидационная: {X_val.shape}")
print(f"   Тестовая: {X_test.shape}")

# ==================== 4. ТЕСТИРОВАНИЕ 5 ОПТИМИЗАТОРОВ ====================
print("\n4. ⚡ Тестирование 5 алгоритмов оптимизации...")

def create_simple_model(input_dim):
    model = Sequential([
        Dense(64, activation='relu', input_shape=(input_dim,)),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    return model

def test_optimizers(X_train, y_train, X_val, y_val):
    optimizers = {
        'Adam': Adam(learning_rate=0.001),
        'SGD': SGD(learning_rate=0.01, momentum=0.9),
        'RMSprop': RMSprop(learning_rate=0.001),
        'Adagrad': Adagrad(learning_rate=0.01),
        'Nadam': Nadam(learning_rate=0.002)
    }

    results = {}

    for opt_name, optimizer in optimizers.items():
        print(f"   Тестируем: {opt_name:<10}", end="")

        model = create_simple_model(X_train.shape[1])
        model.compile(
            optimizer=optimizer,
            loss='binary_crossentropy',
            metrics=['accuracy']
        )

        history = model.fit(
            X_train, y_train,
            validation_data=(X_val, y_val),
            epochs=50,
            batch_size=32,
            verbose=0,
            callbacks=[EarlyStopping(patience=10, restore_best_weights=True)]
        )

        val_loss, val_acc = model.evaluate(X_val, y_val, verbose=0)
        results[opt_name] = {
            'val_accuracy': val_acc,
            'val_loss': val_loss,
            'epochs': len(history.history['loss'])
        }
        print(f" → Точность: {val_acc:.4f}")

    return results

# Тестируем оптимизаторы
optimizer_results = test_optimizers(
    X_train_scaled, y_train,
    X_val_scaled, y_val
)

# ==================== 5. ГРАФИКИ ВИЗУАЛИЗАЦИИ ====================
print("\n5. 📈 Построение 6 графиков визуализации...")

# График 1: Сравнение оптимизаторов
plt.figure(figsize=(10, 6))
names = list(optimizer_results.keys())
accuracies = [optimizer_results[name]['val_accuracy'] for name in names]
colors = ['blue', 'green', 'red', 'purple', 'orange']

plt.subplot(2, 3, 1)
bars = plt.bar(names, accuracies, color=colors)
plt.title('1. Сравнение оптимизаторов')
plt.ylabel('Точность')
plt.ylim([0.5, 1.0])
plt.xticks(rotation=45)
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.01,
             f'{acc:.3f}', ha='center', va='bottom')

# Выбираем лучший оптимизатор
best_opt_name = max(optimizer_results.items(), key=lambda x: x[1]['val_accuracy'])[0]
print(f"\n   🎯 Лучший оптимизатор: {best_opt_name}")

# Создаем финальную модель с лучшим оптимизатором
def create_final_model(input_dim, optimizer_name):
    model = Sequential([
        Dense(128, activation='relu', input_shape=(input_dim,), kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Dropout(0.3),
        Dense(64, activation='relu', kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Dropout(0.2),
        Dense(32, activation='relu'),
        Dropout(0.1),
        Dense(1, activation='sigmoid')
    ])

    if optimizer_name == 'Adam':
        optimizer = Adam(learning_rate=0.0005)
    elif optimizer_name == 'SGD':
        optimizer = SGD(learning_rate=0.01, momentum=0.9)
    elif optimizer_name == 'RMSprop':
        optimizer = RMSprop(learning_rate=0.001)
    elif optimizer_name == 'Adagrad':
        optimizer = Adagrad(learning_rate=0.01)
    else:
        optimizer = Nadam(learning_rate=0.002)

    model.compile(
        optimizer=optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy', 'precision', 'recall', 'AUC']
    )

    return model

# Обучаем финальную модель
print(f"\n6. 🧠 Обучение финальной модели с {best_opt_name}...")
model = create_final_model(X_train_scaled.shape[1], best_opt_name)

callbacks = [
    EarlyStopping(monitor='val_accuracy', patience=30, restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=10),
    ModelCheckpoint('best_ufc_model.h5', save_best_only=True)
]

history = model.fit(
    X_train_scaled, y_train,
    validation_data=(X_val_scaled, y_val),
    epochs=100,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)

# График 2: Точность обучения
plt.subplot(2, 3, 2)
plt.plot(history.history['accuracy'], label='Обучающая')
plt.plot(history.history['val_accuracy'], label='Валидационная')
plt.title('2. Точность модели')
plt.xlabel('Эпоха')
plt.ylabel('Точность')
plt.legend()
plt.grid(True)

# График 3: Потери
plt.subplot(2, 3, 3)
plt.plot(history.history['loss'], label='Обучающая')
plt.plot(history.history['val_loss'], label='Валидационная')
plt.title('3. Потери модели')
plt.xlabel('Эпоха')
plt.ylabel('Потери')
plt.legend()
plt.grid(True)

# Оценка на тестовых данных
print("\n7. 🧪 Тестирование модели...")
test_loss, test_acc, test_precision, test_recall, test_auc = model.evaluate(
    X_test_scaled, y_test, verbose=0
)

# График 4: ROC-кривая
plt.subplot(2, 3, 4)
y_pred_proba = model.predict(X_test_scaled, verbose=0)
fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
roc_auc = auc(fpr, tpr)

plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('4. ROC-кривая')
plt.legend(loc="lower right")
plt.grid(True)

# График 5: Матрица ошибок
plt.subplot(2, 3, 5)
y_pred = (y_pred_proba > 0.5).astype(int)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('5. Матрица ошибок')
plt.xlabel('Предсказано')
plt.ylabel('Истинное')

# График 6: Распределение уверенности
plt.subplot(2, 3, 6)
correct_idx = y_pred.flatten() == y_test
incorrect_idx = y_pred.flatten() != y_test

confidences = np.maximum(y_pred_proba, 1 - y_pred_proba).flatten()
plt.hist(confidences[correct_idx], bins=20, alpha=0.5, label='Правильные', color='green')
plt.hist(confidences[incorrect_idx], bins=20, alpha=0.5, label='Ошибки', color='red')
plt.title('6. Уверенность предсказаний')
plt.xlabel('Уверенность')
plt.ylabel('Частота')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('ufc_visualizations.png', dpi=300, bbox_inches='tight')
plt.show()

# ==================== 8. РЕЗУЛЬТАТЫ ====================
print("\n" + "="*60)
print("🎯 РЕЗУЛЬТАТЫ ТЕСТИРОВАНИЯ:")
print("="*60)
print(f"Оптимизаторы (точность на валидации):")
for name, res in optimizer_results.items():
    print(f"  {name:<10} → {res['val_accuracy']:.4f}")

print(f"\n📊 Результаты на тестовых данных:")
print(f"  Точность:    {test_acc:.4f} ({test_acc*100:.1f}%)")
print(f"  Precision:   {test_precision:.4f}")
print(f"  Recall:      {test_recall:.4f}")
print(f"  AUC:         {test_auc:.4f}")
print(f"  Потери:      {test_loss:.4f}")

print(f"\n📈 Лучший оптимизатор: {best_opt_name}")
print(f"✅ Финальная точность: {test_acc*100:.1f}%")
print("="*60)

# ==================== 9. ПРИМЕРЫ ПРЕДСКАЗАНИЙ ====================
print("\n9. 🔮 Примеры предсказаний:")

np.random.seed(42)
sample_indices = np.random.choice(len(X_test), min(5, len(X_test)), replace=False)

for i, idx in enumerate(sample_indices, 1):
    pred_proba = model.predict(X_test_scaled[idx:idx+1], verbose=0)[0][0]
    true_label = y_test[idx]
    pred_label = 1 if pred_proba > 0.5 else 0

    print(f"\n{i}. Пример #{idx}")
    print(f"   Вероятность победы Fighter_1: {pred_proba:.2%}")
    print(f"   Предсказание: {'Fighter_1' if pred_label == 1 else 'Fighter_2'}")
    print(f"   Истина: {'Fighter_1' if true_label == 1 else 'Fighter_2'}")
    print(f"   {'✅ ПРАВИЛЬНО' if pred_label == true_label else '❌ ОШИБКА'}")

# Сохраняем модель
model.save('ufc_final_model.h5')
print(f"\n💾 Модель сохранена: ufc_final_model.h5")
print(f"📊 Графики сохранены: ufc_visualizations.png")

print(f"\n✅ Готово! Точность модели: {test_acc*100:.1f}%")
```
